# 01. Data Collection

Pulls the core GFDR indicators and WDI controls from the World Bank API using `wbgapi`, and saves a raw country-year panel to `data/raw/`.

**Requires internet access.** Run this notebook locally / in an environment with access to the World Bank API.

```python
import sys
sys.path.append('../src')

import pandas as pd
import wbgapi as wb
from data_utils import INDICATORS, WDI_CONTROLS, fetch_all, add_country_metadata, save_raw
```

## Sanity check: confirm indicator codes resolve

```python
for name, code_ in {**INDICATORS, **WDI_CONTROLS}.items():
    info = wb.series.info(code_)
    print(name, '->', code_)
print(info)
```

## Pull the full panel (2000–2021)

The Dealogic-based syndicated loan / bond series only run 2000–2021, so we bound the pull to that window.

```python
panel = fetch_all(start_year=2000, end_year=2021)
panel = add_country_metadata(panel)
panel.shape
```

## Save raw output

```python
save_raw(panel, '../data/raw/panel_raw.csv')
print('Saved to data/raw/panel_raw.csv')
```
